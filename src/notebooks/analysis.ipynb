```python
from load_data import make_engine
from sqlalchemy import text
```

```python
cohort_query_path = Path.cwd().parent / 'sql/queries/02_cohort.sql'
cohort_query = cohort_query_path.read_text()
```

```python
engine = make_engine()
result_df = pd.read_sql(text(cohort_query), engine)
result_df
```

```python
print(result_df.shape)
print(result_df.dtypes)
```

```python
result_df.loc[result_df['month_index'] == 0, 'people_count'].sum()
```

```python
temp_cohort_df = result_df[result_df['month_index'] == 0]
temp_cohort_df = temp_cohort_df.rename(columns={'people_count': 'cohort_size'})
temp_cohort_df = temp_cohort_df.drop('month_index', axis=1)
temp_cohort_df
```

```python
retention_df = pd.merge(result_df, temp_cohort_df, how='left', on='cohort_month')
retention_df
```

```python
retention_df['retention'] = retention_df['people_count'] / retention_df['cohort_size']
retention_df
```

```python
print(retention_df['cohort_size'].isna().sum())
print((retention_df.loc[retention_df['month_index'] == 0, 'retention'] == 1).all())
```

```python
pivoted_retention_df = retention_df.pivot(index='cohort_month', columns='month_index', values='retention')
pivoted_retention_df
```

```python
pivoted_retention_df.info()
```

```python
pivoted_retention_df = pivoted_retention_df.reindex(columns=range(19))
pivoted_retention_df
```

```python
retention_periods = pd.PeriodIndex(pivoted_retention_df.index, freq='M')
retention_period_end = pd.Period('2018-08', freq='M')
retention_months_to_end = (retention_period_end.year * 12 + retention_period_end.month) - (retention_periods.year * 12 + retention_periods.month)
retention_months_to_end
```

```python
last_observable = retention_months_to_end.to_numpy().reshape(-1, 1)
print(last_observable)

cols = pivoted_retention_df.columns.to_numpy()
print(cols)
```

```python
beyond_range = cols > last_observable
beyond_range
```

```python
empty_cells = pivoted_retention_df.isna()
empty_cells
```

```python
inside = ~beyond_range
inside
```

```python
(empty_cells & inside).sum().sum()
```

```python
filled_df = pivoted_retention_df.fillna(0)
filled_df
```

```python
final_df = filled_df.mask(beyond_range)
final_df
```

```python
print(final_df.notna().sum().sum())
```

```python
final_formatted_df = final_df * 100
final_formatted_df
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
cmap = plt.get_cmap("viridis").copy()
cmap.set_bad("lightgrey")
```

```python
trimmed = final_formatted_df.drop(columns=0)
data = trimmed.to_numpy()
data
```

```python
fig, ax = plt.subplots(figsize=(8, 5))
im = ax.imshow(data, cmap=cmap, aspect="auto")
```

```python
row_labels = trimmed.index.tolist()
col_labels = trimmed.columns.tolist()
```

```python
fig.colorbar(im, ax=ax, label="Retention (%): ordered in that exact month")
ax.set_xticks(range(len(col_labels)))
ax.set_xticklabels(col_labels)
ax.set_yticks(range(len(row_labels)))
ax.set_yticklabels(row_labels)
ax.set_xlabel("Months since first purchase")
ax.set_ylabel("Cohort (month of first delivered order)")
ax.set_title("Cohort Retention: delivered orders only, cohorts Feb 2017 to Aug 2018, and “month 0 not shown”")
```

```python
out_dir = Path("../../reports/figures")
out_dir.mkdir(parents=True, exist_ok=True)
fig.savefig(out_dir / "cohort_retention.png", dpi=150, bbox_inches="tight")
plt.show()
```
